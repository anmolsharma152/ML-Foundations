## **What is Reinforcement Learning**?

Reinforcement Learning (RL) is a branch of Machine Learning where an agent learns by interacting with its environment.

Instead of learning from **labeled examples**, the **agent learns** through:

- Trial and Error
- Rewards
- Penalties

**Example**

**Imagine teaching a dog:**

- Sit correctly → Give a treat (+ reward)
Ignore command → No reward

- After many attempts, the dog learns the correct behavior.

Similarly, an **RL agent learns the best actions**.

## **5. What is Q-Learning?**

Q-Learning is one of the most popular Reinforcement Learning algorithms.

The letter **Q stands** for: **Quality**

It measures: **How good is an action** in a **particular state?**

## **6. Q-Table**

The agent **stores knowledge** in a **Q-table**.

**Example**:

| State | Left | Right |
| ----- | ---- | ----- |
| S0    | 0    | 5     |
| S1    | 2    | 7     |
| S2    | 1    | 10    |


**What Does Q-value = 10 Mean?**

Suppose: S0 → S1 → S2 → Goal and reaching the Goal gives: **Reward = +10**

After many training episodes, the agent learns:

When I am at S2,
moving Right takes me directly to the Goal.

**Therefore**:

Q(S2, Right) = 10 becomes very high.

The **highest value** indicates the **best action**.

**Q-values work exactly the same way:**

- Higher Q-value = Better action = More likely to reach the goal

The **agent has learned that moving Right** from every state is the **optimal strategy to maximize future rewards**.

**7. Q-Learning Formula**

The **Q-value** is **updated** using:

Q(s,a)=Q(s,a)+α[R+γmax a Q(s′,a)−Q(s,a)]

**Parameters**

| Symbol | Meaning         |
| ------ | --------------- |
| Q(s,a) | Current Q-value |
| α      | Learning rate   |
| γ      | Discount factor |
| R      | Reward          |
| s'     | Next state      |

## **11. Training the Agent**

```python
import numpy as np
import random

# Define Environment
num_states = 5    # 5 states
num_actions = 2   # 2 possible action

# Create Q-Table
q_table = np.zeros((num_states, num_actions))

# Learning Parameters
alpha = 0.1  # agent learns 10% from new experiences.
gamma = 0.9 # I am willing to wait for a bigger reward in the future.

episodes = 1000   # Train the agent 1000 times.

for episode in range(episodes):

    state = 0

    while state < 4:             # Keep moving until reaching S4. and LOOP STOP state == 4.

        action = random.randint(0, 1)      #  Choose a random action.
        #print("action", random.randint(0, 1))

        # calculate next state
        if action == 1:                   # means right, Move one step forward.
            next_state = min(state + 1, 4)   # why min ?  4+1 = 5 i.e s5 not exist
        else:
            next_state = max(state - 1, 0)    # move left   0-1=-1(invalid state) so stay at s0

        reward = 10 if next_state == 4 else -1     # Assign Reward

        old_q = q_table[state, action] # Current Q-Value

        next_max = np.max(q_table[next_state])   # Best Future Value

        new_q = old_q + alpha * (
            reward + gamma * next_max - old_q    # Update Q-Value
        )

        q_table[state, action] = new_q   # Save New Q-Value (Store updated value back into the Q-table.)

        state = next_state # agent moves to the next state.
```

## **12. View Learned Q-Table**

```python
print(q_table)
```

The values for moving right become larger because moving right eventually reaches the goal.

## **13. Testing the Agent**

After training, we **allow the agent** to **use** the **best action**.

```python
state = 0

print("Path:")

while state < 4:

    action = np.argmax(q_table[state])

    if action == 1:
        state += 1
    else:
        state -= 1

    print(state)
```

The **agent** successfully **reaches the goal**.

## **16. Exploration vs Exploitation**

**Exploration** - Trying new actions.

```python
action = random.randint(0, 1)
```
