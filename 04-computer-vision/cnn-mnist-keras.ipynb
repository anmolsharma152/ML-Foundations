```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, MaxPooling2D, Input

# --- 1. DATA PREPARATION ---
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# NORMALIZE (0 to 1)
X_train = X_train / 255.0
X_test = X_test / 255.0

# RESHAPE: CNNs need shape (28, 28, 1) -> The '1' is the grayscale channel.
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print(f"Data Reshaped. X_train shape: {X_train.shape}")

# --- 2. BUILD MODEL (THE CNN ARCHITECTURE) ---
model = Sequential([
    Input(shape=(28, 28, 1)),

    # CONV LAYER 1: The "Eye"
    # It scans the image with 32 different filters (3x3 size).
    # It creates 32 "Feature Maps" (maps of where edges/curves are).
    Conv2D(32, kernel_size=(3, 3), activation='relu'),

    # POOLING LAYER 1: The "Compressor"
    # It shrinks the image by 2x (keeping only the strongest features).
    # This reduces calculation time and focuses on the important bits.
    MaxPooling2D(pool_size=(2, 2)),

    # CONV LAYER 2: The "Brain"
    # It looks at the features found by Layer 1 and combines them.
    # e.g., "Vertical line" + "Horizontal line" = "Corner"
    Conv2D(64, (3, 3), activation='relu'),

    # POOLING LAYER 2
    MaxPooling2D(pool_size=(2, 2)),

    # NOW we flatten.
    # But now, we aren't flattening raw pixels.
    # We are flattening "High Level Features" (like "Circle detected at top").
    Flatten(),

    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# --- 3. TRAIN ---
print("\nStarting Training on CNN...")
model.fit(X_train, y_train, epochs=5)

# --- 4. EVALUATE ---
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nTest Accuracy: {test_acc*100:.2f}%")
```
