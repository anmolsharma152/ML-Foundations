```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# --- PHASE 1: PREPARING REAL DATA ---

def is_prime(n):
    if n < 2: return 0
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return 0
    return 1

# Generate numbers 1 to 10,000
numbers = np.array(range(1, 10001))

# Create the labels (The "Teacher's Answer Key")
# This creates a list of 0s and 1s corresponding to each number
labels = np.array([is_prime(n) for n in numbers])

# SCALING: Neural networks struggle with large numbers like "9000".
# We divide by 10,000 to keep inputs between 0 and 1.
X = numbers / 10000.0
y = labels

print(f"Data Prepared: {len(X)} numbers.")
print(f"Example: Number {numbers[6]} (Scaled: {X[6]}) is Prime? {y[6]}")

# --- PHASE 2: BUILDING THE MODEL (The Architecture) ---

model = Sequential([
    # We use Input(shape=(1,)) to fix the warning from your last run
    Input(shape=(1,)),
    Dense(64, activation='relu'),  # Hidden layer trying to find patterns
    Dense(32, activation='relu'),  # Another layer for complexity
    Dense(1, activation='sigmoid') # Output: Probability of being Prime
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# --- PHASE 3: TRAINING (The Learning) ---

print("\nStarting Training on Prime Numbers...")
# We use validation_split to hold back 20% of data to test on unseen numbers
model.fit(X, y, epochs=10, batch_size=32, validation_split=0.2, verbose=1)

# --- PHASE 4: PREDICTION (The Test) ---
# Let's ask the model about specific numbers it may not have memorized perfectly.

print("\n--- PREDICTIONS ---")
test_numbers = np.array([17, 20, 997, 4000]) # 17=Prime, 20=Not, 997=Prime
scaled_tests = test_numbers / 10000.0

predictions = model.predict(scaled_tests)

for num, pred in zip(test_numbers, predictions):
    # If prediction > 0.5, the model thinks it is Prime
    status = "Prime" if pred > 0.5 else "Not Prime"
    confidence = pred[0] * 100
    print(f"Number: {num:<5} | Model Guessed: {status} ({confidence:.2f}%)")
```
