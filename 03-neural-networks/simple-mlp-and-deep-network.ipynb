```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. DEFINE THE NETWORK (MLP)
class SimpleMLP(nn.Module):
    def __init__(self):
        super(SimpleMLP, self).__init__()
        # Input layer (2 features) -> Hidden layer (4 neurons)
        self.hidden = nn.Linear(2, 4)
        # Activation function
        self.relu = nn.ReLU()
        # Hidden layer -> Output layer (1 prediction)
        self.output = nn.Linear(4, 1)

    def forward(self, x):
        # Pass data through hidden layer
        x = self.hidden(x)
        # Apply ReLU activation
        x = self.relu(x)
        # Pass data to output layer
        x = self.output(x)
        return x

# Initialize model
model = SimpleMLP()

# 2. CREATE DUMMY DATA
# Input: [2.0, 3.0], Target: [10.0]
inputs = torch.tensor([[2.0, 3.0]])
target = torch.tensor([[10.0]])

# 3. DEFINE LOSS AND OPTIMIZER
criterion = nn.MSELoss()  # Mean Squared Error
optimizer = optim.SGD(model.parameters(), lr=0.01) # Stochastic Gradient Descent

print(f"Initial weights (partial): {model.hidden.weight[0][:2]}")

# 4. TRAINING LOOP (The Cycle)
for epoch in range(5):
    # Step A: Forward Pass (Predict)
    prediction = model(inputs)

    # Step B: Calculate Loss (Error)
    loss = criterion(prediction, target)

    # Step C: Backward Pass (Backpropagation)
    # This single line calculates ALL gradients using the Chain Rule automatically
    optimizer.zero_grad() # Clear old gradients
    loss.backward()

    # Step D: Update Weights
    optimizer.step()

    print(f"Epoch {epoch+1} | Loss: {loss.item():.4f} | Prediction: {prediction.item():.4f}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Data
inputs = torch.tensor([[2.0, 3.0]])
target = torch.tensor([[10.0]])

def train_model(learning_rate, iterations=5):
    # Re-initialize model every time to be fair
    model = nn.Sequential(nn.Linear(2, 1))

    # Initialize weights to a fixed value for consistent comparison
    with torch.no_grad():
        model[0].weight.fill_(0.5)
        model[0].bias.fill_(0.0)

    optimizer = optim.SGD(model.parameters(), lr=learning_rate)
    criterion = nn.MSELoss()

    print(f"\n--- Training with LR = {learning_rate} ---")
    for epoch in range(iterations):
        prediction = model(inputs)
        loss = criterion(prediction, target)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        print(f"Epoch {epoch+1}: Loss = {loss.item():.4f}")

# Experiment 1: Small Step (Safe but slow)
train_model(learning_rate=0.001)

# Experiment 2: Moderate Step (Fast convergence)
train_model(learning_rate=0.01)

# Experiment 3: Huge Step (Overshooting / Exploding)
train_model(learning_rate=1.5)
```

```python
class DeepNetwork(nn.Module):
    def __init__(self):
        super(DeepNetwork, self).__init__()

        self.layers = nn.Sequential(
            # Layer 1: Input (2 features) -> Hidden (10 neurons)
            nn.Linear(2, 10),
            nn.ReLU(),

            # Layer 2: Hidden (10) -> Hidden (10)
            # This "extra" layer is what makes it "Deep"
            nn.Linear(10, 10),
            nn.ReLU(),

            # Layer 3: Hidden (10) -> Output (1 prediction)
            nn.Linear(10, 1)
        )

    def forward(self, x):
        return self.layers(x)

# Setup
deep_model = DeepNetwork()
print(deep_model)
```
