```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("heesoo37/120-years-of-olympic-history-athletes-and-results")

print("Path to dataset files:", path)
```

```python
path
```

```python
import os
import pandas as pd

# Listing the downloaded data
files = os.listdir(path)

# Creating the complete path
path1 = os.path.join(path, files[0])
path2 = os.path.join(path, files[1])
```

```python
df_athletes = pd.read_csv(path1)
df_noc = pd.read_csv(path2)
```
