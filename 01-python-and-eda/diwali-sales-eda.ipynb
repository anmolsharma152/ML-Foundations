```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("Diwali Sales Data.csv",encoding='unicode_escape')
```

```python
# Removing Null/Blank Columns
df = df.drop(columns=['Status','unnamed1'])
```

```python
# Removing null records from amount column:
df.dropna(inplace=True)
```

```python
df
```

```python
df.isnull().sum()
```

```python
# Removing duplicates from dataset:
print("Duplictes before deletion: ",df.duplicated().sum())

df = df.drop_duplicates()

print("Duplictes after deletion: ",df.duplicated().sum())
```

```python
## Gender Distribution:
plt.figure(figsize=(10,5))

ax = sns.countplot(x = 'Gender', data = df)
for bars in ax.containers:
    ax.bar_label(bars)

plt.title("Gender Distribution based on Orders")
plt.show()
```

```python
gender_amount_df = pd.DataFrame(df.groupby('Gender', as_index=False)['Amount'].sum())
gender_amount_df = gender_amount_df.sort_values(by='Amount', ascending=False)

plt.figure(figsize=(10,5))

ax = sns.barplot(x='Gender', y='Amount', data=gender_amount_df)

plt.title("Gender Distribution Based on Amount Spent")
plt.show()
```
